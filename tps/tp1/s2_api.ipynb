## Utiliser Wikidata pour récupérer les différents appellations d'un concept

https://www.wikidata.org

Objectif : enrichir CAMille en récupérant les noms alternatifs d'un concept, afin de pouvoir le rechercher plus efficacement dans le corpus.

```python
concept = "Dyslexie"
# Récupération des noms et alias depuis Wikidata
headers = {"User-Agent": "TP-TAC-ULB (mireiaparetsperis@ulb.be)"}
url = "https://www.wikidata.org/w/api.php"
data = {
    "action": "wbgetentities",
    "sites": "frwiki",
    "titles": concept,
    "props": "labels|descriptions|aliases",
    "format": "json",
    "languages": "fr"
}
resp = requests.get(url, params=data, headers=headers)
try:
    json_list = json.loads(resp.text)
except json.JSONDecodeError: # réponse probablement bloquée
    print(resp.text)
for item in json_list["entities"].values():
    label = item["labels"]["fr"]["value"]
    description = item["descriptions"]["fr"]["value"]
    print(f"{label} ({description})")
    for alias in item["aliases"].get("fr", []):
        print(alias["value"])
```
