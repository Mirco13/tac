# APIs: requêtes HTTP

## Imports

```python
import json
import requests
```

## Utiliser Wikidata pour récupérer les différentes appellations d'un concept

https://www.wikidata.org

Objectif : enrichir CAMille en récupérant les noms alternatifs d'un concept, afin de pouvoir le rechercher plus efficacement dans le corpus.

```python
concept = "Dyslexie|Dysorthographie|Dysgraphie|Belgique"
# Récupération des noms et alias depuis Wikidata
headers = {"User-Agent": "TP-TAC-ULB (mireiaparetsperis@ulb.be)"}
url = "https://www.wikidata.org/w/api.php"
data = {
    "action": "wbgetentities",
    "sites": "frwiki",
    "titles": concept,
    "props": "labels|descriptions|aliases",
    "format": "json",
    "languages": "fr"
}
resp = requests.get(url, params=data, headers=headers)
try:
    json_list = json.loads(resp.text)
except json.JSONDecodeError: # réponse probablement bloquée
    print(resp.text)
for item in json_list["entities"].values():
    label = item["labels"]["fr"]["value"]
    description = item["descriptions"]["fr"]["value"]
    print(f"{label} ({description})")
    for alias in item["aliases"].get("fr", []):
        print(alias["value"])
```

## Conclusion

Cette requête permet de récupérer le nom actuel et une description des termes liés aux troubles de l'apprentissage de la lecture et de l'écriture. Wikidata ne fournit aucune appellation historique (comme "cécité verbale"), mais seulement quelques variantes orthographiques et morphologiques (comme "disorthographique"). Ces termes devront être complétés par d'autres sources pour rechercher ces notions dans CAMille.

Le terme de contrôle "Belgique", pour lequel l'API renvoie plusieurs alias, confirme que la requête fonctionne correctement.
