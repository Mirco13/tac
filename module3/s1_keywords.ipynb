# Extraction de Keywords

## Imports

```python
import os
import yake
from collections import Counter, defaultdict
```

## Extraire les mots clés d'un document avec Yake

https://github.com/INESCTEC/yake

```python
# Instantier l'extracteur de mots clés
kw_extractor = yake.KeywordExtractor(lan="fr", top=50)
kw_extractor
```

```python
# Lister les Fichiers
data_path = "../data/txt/"
files = [f for f in os.listdir(data_path) if f.endswith('.txt')]
```

```python
# Imprimer le nombre de fichiers identifiés
len(files)
```

```python
# Nombre de fichiers par année (pour choisir l'année)
count_year = defaultdict(int)
for f in files:
    y = f.split("_")[2].split("-")[0]
    count_year[y] += 1
for y in sorted(count_year):
    print(y, count_year[y])
```

```python
# Recherche de termes liés à l'école et à la lecture
terms = ["dyslexi", "cécité verbale", "decroly", "scolaire", "lecture", "école",
         "enseignement", "apprentissage", "alphabétisation"]
counts = {}
for t in terms:
    counts[t] = defaultdict(int)
for f in files:
    y = f.split("_")[2].split("-")[0]
    text = open(os.path.join(data_path, f), 'r', encoding='utf-8').read().lower()
    for t in terms:
        counts[t][y] += text.count(t)
```

```python
# Trier et imprimer les cinq années principales pour chaque terme 
for t in terms:
    sorted_years = sorted(counts[t].items(), key=lambda kv: kv[1], reverse=True)
    total = sum(counts[t].values())
    print(f"{t} : {total} au total | années principales : {sorted_years[:5]}")
```

```python
# Les dix premiers fichiers
files[:10]
```

```python
# Choisir une année pour l'extraction de mots clés
year = "1914"

# ne garder que les fichiers de l'année choisie
files_year = sorted(f for f in files if f.split("_")[2].split("-")[0] == str(year))
len(files_year)
```

```python
# Vérification pour 1914 : mentions de la loi scolaire et de la guerre
for t in ["poullet", "loi scolaire", "instruction obligatoire", "enseignement primaire", "guerre", "question scolaire"]:
    n = 0
    for f in files_year:
        text = open(os.path.join(data_path, f), 'r', encoding='utf-8').read().lower()
        n += text.count(t)
    print(f"{t} : {n}")
```

```python
# Choisir un fichier
this_file = files_year[0]
this_file
```

```python
# Récupérer le texte du fichier
text = open(os.path.join(data_path, this_file), 'r', encoding='utf-8').read()
text[:500]
```

```python
# Extraire les mots clés de ce texte
keywords = kw_extractor.extract_keywords(text)
keywords
```

```python
# Ne garder que les bigrammes
kept = []
for kw, score in keywords:
    words = kw.split()
    if len(words) == 2:
        kept.append(kw)
kept
```

## Faire la même opération sur tous les documents

```python
for f in files_year:
    text = open(os.path.join(data_path, f), 'r', encoding="utf-8").read()
    keywords = kw_extractor.extract_keywords(text)
    kept = []
    for kw, score in keywords:
        words = kw.split()
        if len(words) == 2:
            kept.append(kw)
    print(f"{f} mentions these keywords: {', '.join(kept)}...")
```

```python
# Mots-clés les plus fréquents dans l'année choisie
all_keywords = Counter()
for f in files_year:
    text = open(os.path.join(data_path, f), 'r', encoding="utf-8").read()
    keywords = kw_extractor.extract_keywords(text)
    for kw, score in keywords:
        if len(kw.split()) == 2:
            all_keywords[kw.lower()] += 1
all_keywords.most_common(30)
```

## Répartition du mot "guerre" dans l'année

```python
# Compter le nombre "guerre" dans chaque page de l'année.
guerre_par_page = defaultdict(int)
for f in files_year:
    text = open(os.path.join(data_path, f), 'r', encoding="utf-8").read().lower()
    guerre_par_page[f] = text.count("guerre")   
```

```python
# Trier et imprimer les 10 pages qui contiennent le plus
sorted_pages = sorted(guerre_par_page.items(), key=lambda kv: kv[1], reverse=True)

for page, freq in sorted_pages[:10]:
    print(f"{page} : {freq} fois le mot 'guerre'")  
```
