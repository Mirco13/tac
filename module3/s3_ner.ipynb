```python
# Appliquer le test sur toutes les phrases
for sent in sentences:
    doc = nlp(sent)
    entities = []
    for ent in doc.ents:
        entities.append(f"{ent.text} ({ent.label_})")
    if entities:
        print(f"'{doc.text}' contient les entités suivantes : {', '.join(entities)}")
    else:
        print(f"'{doc.text}' ne contient aucune entité")
```

## Appliquer la reconnaissance d'entités nommées sur notre corpus

```python
# Lister les fichiers de l'année 1914
year = 1914
data_path = '../data/txt'
files_year = [f for f in os.listdir(data_path) if f.split("_")[2].split("-")[0] == str(year)] 
len(files_year)
```

```python
%%time 
# Traiter chaque page de l'année avec spaCy 
# (page par page, car spaCy refuse les textes de plus d'un million de caractères) 
docs = [] 
for f in files_year: 
    text = open(os.path.join(data_path, f), 'r', encoding='utf-8').read() 
    docs.append(nlp(text))
```

```python
# Compter les entités 
people = defaultdict(int) 
for doc in docs: 
    for ent in doc.ents: 
        if ent.label_ == "PER" and len(ent.text) > 3: 
            people[ent.text] += 1
```

```python
# Trier et imprimer

sorted_people = sorted(people.items(), key=lambda kv: kv[1], reverse=True)

for person, freq in sorted_people[:50]:
    print(f"{person} apparait {freq} fois dans le corpus")
```

Exercice: essayez de lister les lieux (LOC) et les organisations (ORG) les plus mentionnées dans le corpus

```python
# Compter les lieux
places = defaultdict(int)
for doc in docs:
    for ent in doc.ents:
        if ent.label_ == "LOC" and len(ent.text) > 3:
            places[ent.text] += 1
```

```python
# Trier et imprimer les lieux
sorted_places = sorted(places.items(), key=lambda kv: kv[1], reverse=True) 
for place, freq in sorted_places[:50]: 
    print(f"{place} apparait {freq} fois dans le corpus")
```

```python
# Compter les organisations
orgs = defaultdict(int) 
for doc in docs: 
    for ent in doc.ents: 
        if ent.label_ == "ORG" and len(ent.text) > 3: 
            orgs[ent.text] += 1
```

```python
# Trier et imprimer les organisations 
sorted_orgs = sorted(orgs.items(), key=lambda kv: kv[1], reverse=True) 

for org, freq in sorted_orgs[:50]: 
    print(f"{org} apparait {freq} fois dans le corpus")
```
